## 📌 Objectives
1. Compute global and local feature attribution using **SHAP TreeExplainer**.
2. Decouple **controllable management factors** ($N, P, K$, pH, SOM) from **exogenous weather**.
3. Validate actionable intervention thresholds for the Prescriptive Farmer Advisory Script.

```python
import joblib
import pandas as pd
import shap
import matplotlib.pyplot as plt

# Load model and feature store
checkpoint = joblib.load('../models/best_yield_regressor.joblib')
model = checkpoint['model']
features = checkpoint['feature_names']

df = pd.read_csv('../data/processed/train_features.csv')
X = df[features]
print(f"Loaded feature matrix: {X.shape}")

# TreeExplainer attribution
explainer = shap.TreeExplainer(model)
shap_values = explainer.shap_values(X)
print("SHAP attribution complete!")
```

## 1. Global SHAP Summary Beeswarm Plot

```python
plt.figure(figsize=(10, 6))
shap.summary_plot(shap_values, X, max_display=12)
plt.show()
```
