```python
import pandas as pd
import numpy as np
import joblib
from sklearn.metrics import mean_squared_error, r2_score

# Load engineered feature store
df = pd.read_csv('../data/processed/train_features.csv')
print(f'Training Feature Store: {df.shape[0]} rows x {df.shape[1]} columns')

# Load production model
checkpoint = joblib.load('../models/best_yield_regressor.joblib')
print(f"Best Production Model: {checkpoint['model_name']}")
print(f"Spatial CV Metrics: {checkpoint['spatial_metrics']}")
```

## 1. Actual vs. Predicted Parity Plot Across Unseen Districts

```python
import matplotlib.pyplot as plt
import seaborn as sns

features = checkpoint['feature_names']
model = checkpoint['model']
X = df[features]
y = df['crop_yield_mt_ha']
preds = model.predict(X)

fig, ax = plt.subplots(figsize=(8, 8))
ax.scatter(y, preds, alpha=0.7, c='navy')
ax.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')
ax.set_title(f"Model Goodness-of-Fit (R2 = {r2_score(y, preds):.3f})")
ax.set_xlabel('Actual Yield MT/Ha')
ax.set_ylabel('Fitted Yield MT/Ha')
plt.show()
```
