## 📌 Objectives
1. Explore stoichiometric nutrient ratios ($N:P$, $N:K$, $P:K$).
2. Investigate hydrothermal aridity indices and thermal stress penalties.
3. Quantify feature-target linear and non-linear correlations with end-of-season crop yield.
4. Validate feature store integrity for Spatial Block Cross-Validation.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')
df = pd.read_csv('../data/processed/train_features.csv')
print(f'Feature Store Loaded: {df.shape[0]} rows x {df.shape[1]} columns')
df[['crop_type', 'soil_type', 'crop_yield_mt_ha', 'n_to_p_ratio', 'hydro_thermal_stress_index']].head()
```

## 1. Stoichiometric Response: N:P Ratio vs. Yield

```python
fig, ax = plt.subplots(figsize=(10, 5))
sns.scatterplot(data=df, x='n_to_p_ratio', y='crop_yield_mt_ha', hue='crop_type', s=70, ax=ax)
sns.regplot(data=df, x='n_to_p_ratio', y='crop_yield_mt_ha', scatter=False, color='red', ax=ax)
ax.set_title('Crop Yield as a Function of Nitrogen-to-Phosphorus Stoichiometry', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 2. Hydrothermal Stress Landscape: Moisture vs. Evaporative Demand

```python
fig, ax = plt.subplots(figsize=(10, 5))
sc = ax.scatter(df['mean_vpd_kpa'], df['hydro_thermal_stress_index'], c=df['crop_yield_mt_ha'], cmap='viridis', s=70)
plt.colorbar(sc, label='Crop Yield (MT/Ha)')
ax.set_xlabel('Vapor Pressure Deficit (kPa)')
ax.set_ylabel('Hydrothermal Stress Index')
ax.set_title('Yield Distribution across Climate Stress Space', fontweight='bold')
plt.tight_layout()
plt.show()
```

## 3. Pearson Correlation Ranking across All Engineered Features

```python
numeric_df = df.select_dtypes(include=[np.number])
corrs = numeric_df.corr()['crop_yield_mt_ha'].drop(['crop_yield_mt_ha', 'production_mt'], errors='ignore')
top_corrs = corrs.abs().sort_values(ascending=False).head(15)
print('Top 15 Most Correlated Features with Crop Yield:')
print(corrs.loc[top_corrs.index])
```
